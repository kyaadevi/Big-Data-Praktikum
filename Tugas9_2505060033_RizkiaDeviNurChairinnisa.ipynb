**B. Kueri Bertingkat (Read)** *(bobot 20%)*

Tampilkan hasil dari **tiga** kueri berikut :

1. Seluruh menu kategori "Kopi" dengan `tingkat_kepahitan` di atas 3.

```python
print("kategori kopi dengan tingkat kepahitan di atas 3:")
for p in koleksi_menu.find({"kategori": "coffee", "tingkat_kepahitan": {"$gt": 3}}):
    print(f" > {p['nama_menu']}, tingkat kepahitan: {p['tingkat_kepahitan']}")
```

2. Seluruh menu dengan `harga` di antara 15000 dan 30000.

```python
print("SELURUH menu dengan harga lebih dari 15000 kurang dari 30000")
# Jadi dari 16000 sampai 29000 
for p in koleksi_menu.find({
    "harga": {
        "$gt": 15000,
        "$lt": 30000}}):
    print(f" - {p['nama_menu']} (Rp{p['harga']:,})")
```

3. Menu yang **tidak memiliki** field `pedas_level` sama sekali.

```python
print("menu yang tidak memiliki field pedas_level:")
for p in koleksi_menu.find({"pedas_level": {"$exists": False}}):
    print(f" - {p['nama_menu']}, kategori: {p['kategori']}")
```

**D. Integrasi ke Spark** *(bobot 25%)*

Ambil seluruh data dari collection `menu`, ubah menjadi Spark DataFrame (ikuti pola Sub-bab 9.6 — perhatikan field yang mungkin kosong/`null` pada sebagian dokumen, seperti `tingkat_kepahitan` yang hanya dimiliki menu Kopi). Tampilkan:
1. `printSchema()` hasil konversi.

```python
from pyspark.sql import SparkSession
import pandas as pd

spark = SparkSession.builder \
    .appName("Tugas9-MongoDBSpark") \
    .master("local[*]") \
    .getOrCreate()
spark.sparkContext.setLogLevel("ERROR")

dokumen_mongo =list(koleksi_menu.find({}))

df_pandas = pd.DataFrame(dokumen_mongo)
df_pandas["_id"] = df_pandas["_id"].astype(str)

df_spark_menu = spark.createDataFrame(df_pandas)

df_spark_menu.printSchema()
df_spark_menu.show()
```

2. Rata-rata `harga` per `kategori` menggunakan `groupBy`.

```python
from pyspark.sql.functions import avg, count

df_spark_menu.groupBy("kategori").agg(
    avg("harga").alias("rata-rata_harga"),
    count("nama_menu").alias("jumlah_menu")
).show()
```

**E. Kesimpulan** *(bobot 20%)*

Tulis pada markdown cell (**minimal 100 kata**): apa tantangan yang anda temui saat mengonversi data MongoDB (dengan field yang tidak seragam antar dokumen) menjadi Spark DataFrame (yang mengharuskan skema tabel seragam)? Bagaimana anda mengatasinya?

**Jawaban**:

Tantangan utama saat mengkonversi data dari MongoDB ke Spark DataFrame yaitu pada perbedaan cara mereka menyimpan data. MongoDB lebih ke fleksibel, setiap dokumen bisa punya kolom yang berbeda-beda.
Contoh: menu kopi memiliki `tingkat_kepahitan`, sedangkan menu makanan memiliki `pedas_level`
Disis lain, Spark DataFrame itu kaku dan harus semua baris data memiliki struktur kolom yang sama persis.
Akibatnya, ketika data yang tidak setipe ini dimasukkan ke dalam Spark, Spark akan otomatis mengisi dengan nilai NaaN (kosong) pada kolom yang tidak dimiliki oleh suatu menu. Terlihat jelas pada hasil di mana kolom `tingkat_kepahitan` berisi kosong untuk menu makanan.

Untuk mengatasinya, saya melakukan penyesuaian tipe data agar Spark bisa membacanya dengan benar.Misalnya, mengubah format `_id` dari mongoDB menjadi teks (String) dan membersihkan data yang formatnya tidak sesuai. Selain itu, saat melakukan `groupBy` untuk mencari rata-rata harga, fungsi agregasi di Spark secara otomatis mengabaikan nilai NaN. Dengan penyesuaian ini, proses integrasi data tetap berjalan lancar dan hasil analisinya tetap akurat, meskipun data aslinya tidak sama.
